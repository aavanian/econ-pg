## 1. Basic Market Equilibrium

Let's create a simple market and find its equilibrium.

```python
# Create a market with default parameters
market = Market(
    demand_intercept=100,
    demand_slope=2,
    supply_intercept=20,
    supply_slope=1.5
)

print(market)
```

```python
# Visualize the market
plot_supply_demand(market)
```

## 2. Economic Interpretation

### Equilibrium Point
The equilibrium occurs where supply equals demand. At this point:
- The market clears (no excess supply or demand)
- Both consumers and producers are satisfied
- Price signals balance production and consumption decisions

### Consumer Surplus
The blue shaded area represents consumer surplus - the benefit consumers receive from paying less than their maximum willingness to pay.

### Producer Surplus
The red shaded area represents producer surplus - the benefit producers receive from selling at a price higher than their minimum acceptable price.

## 4. Comparative Statics

How does equilibrium change when we shift demand or supply?

```python
# Create a table showing how equilibrium changes with demand shifts
import pandas as pd

results = []
for demand_intercept in range(80, 141, 10):
    m = Market(
        demand_intercept=demand_intercept,
        demand_slope=2,
        supply_intercept=20,
        supply_slope=1.5
    )
    price, quantity = m.equilibrium()
    results.append({
        'Demand Intercept': demand_intercept,
        'Equilibrium Price': f'{price:.2f}',
        'Equilibrium Quantity': f'{quantity:.2f}',
        'Consumer Surplus': f'{m.consumer_surplus():.2f}',
        'Producer Surplus': f'{m.producer_surplus():.2f}'
    })

df = pd.DataFrame(results)
print("\nEffect of Demand Shifts on Equilibrium:")
print(df.to_string(index=False))
```

## 5. Elasticity Analysis

Let's examine how the slope parameters (elasticities) affect the market.

```python
fig, axes = plt.subplots(1, 2, figsize=(15, 5))

# Left: Effect of demand elasticity
ax1 = axes[0]
quantities = np.linspace(0, 50, 100)

for slope, label in [(1, 'Elastic'), (2, 'Moderate'), (4, 'Inelastic')]:
    m = Market(demand_intercept=100, demand_slope=slope, supply_intercept=20, supply_slope=1.5)
    prices = m.demand_price(quantities)
    eq_p, eq_q = m.equilibrium()
    
    ax1.plot(quantities, prices, label=f'{label} (slope={slope})', linewidth=2)
    ax1.plot(eq_q, eq_p, 'o', markersize=8)

# Add supply curve
supply_m = Market()
ax1.plot(quantities, supply_m.supply_price(quantities), 'r--', label='Supply', linewidth=2)

ax1.set_xlabel('Quantity')
ax1.set_ylabel('Price')
ax1.set_title('Effect of Demand Elasticity', fontweight='bold')
ax1.legend()
ax1.grid(True, alpha=0.3)
ax1.set_xlim(0, 50)

# Right: Effect of supply elasticity
ax2 = axes[1]

for slope, label in [(0.5, 'Elastic'), (1.5, 'Moderate'), (3, 'Inelastic')]:
    m = Market(demand_intercept=100, demand_slope=2, supply_intercept=20, supply_slope=slope)
    prices = m.supply_price(quantities)
    eq_p, eq_q = m.equilibrium()
    
    ax2.plot(quantities, prices, label=f'{label} (slope={slope})', linewidth=2)
    ax2.plot(eq_q, eq_p, 'o', markersize=8)

# Add demand curve
demand_m = Market()
ax2.plot(quantities, demand_m.demand_price(quantities), 'b--', label='Demand', linewidth=2)

ax2.set_xlabel('Quantity')
ax2.set_ylabel('Price')
ax2.set_title('Effect of Supply Elasticity', fontweight='bold')
ax2.legend()
ax2.grid(True, alpha=0.3)
ax2.set_xlim(0, 50)

plt.tight_layout()
plt.show()
```
